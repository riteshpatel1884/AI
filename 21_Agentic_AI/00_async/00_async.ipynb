### Sync Programming

```python
import time
```

```python
def fetch_data(name):
    print(f"Starting {name}...")
    time.sleep(2)  # Simulates API/LLM call
    print(f"Finished {name}")
    return f"{name} result"


start = time.time()

result1 = fetch_data("Web Search")
result2 = fetch_data("Database")
result3 = fetch_data("News API")

print(f"Total time: {time.time() - start:.2f} seconds")
```

fetch_data("Web Search") --> sleep 2 sec --> return result1 --> fetch_data("Database") --> sleep 2 sec --> return result2 --> fetch_data("News API") --> sleep 2 sec --> return result3

```python
import asyncio
import time

async def fetch_data(name):
    print(f"Starting {name}...")
    await asyncio.sleep(2)  # Simulates API/LLM call. Use asyncio.sleep instead of time.sleep in case if u are using await.
    print(f"Finished {name}")
    return f"{name} result"


async def main():
    start = time.time()
    results = await asyncio.gather(
        fetch_data("Web Search"),
        fetch_data("Database"),
        fetch_data("News API")
    )

    print(results)
    print(f"Total time: {time.time() - start:.2f} seconds")


await main()
```
